### Exercise 2
What are the quotient and remainder of 29 divided by 5?

```python
# Exercise 2: Quotient and remainder
number = 29
divisor = 5

quotient = number // divisor
remainder = number % divisor

print("Quotient:", quotient)
print("Remainder:", remainder)
```

## Exercises: Lists

### Exercise 1: Create a List
Create a list called `colors` containing `"red", "green", "blue", "yellow"`. Print the list.

```python
# Exercise 1: Create a list
colors = ["red", "green", "blue", "yellow"]
print(colors)
```

### Exercise 2: Access Elements
Using `animals = ["cat", "dog", "rabbit", "horse", "bird"]`:
1. Print the first element.
2. Print the third element.
3. Print the last element.

```python
# Exercise 2: Access elements
animals = ["cat", "dog", "rabbit", "horse", "bird"]

print("First:", animals[0])
print("Third:", animals[2])
print("Last:", animals[-1])
```

### Exercise 3: Negative Indexing
Using the same `animals` list, print:
1. The last element.
2. The second-to-last element.
3. The third-to-last element.

```python
# Exercise 3: Negative indexing
print("Last:", animals[-1])
print("Second-to-last:", animals[-2])
print("Third-to-last:", animals[-3])
```

### Exercise 4: Basic Slicing
Given `numbers = [10, 20, 30, 40, 50, 60]`, create slices containing:
1. The first three elements.
2. The last three elements.
3. The middle two elements.

```python
# Exercise 4: Basic slicing
numbers = [10, 20, 30, 40, 50, 60]

first_three = numbers[:3]
last_three = numbers[-3:]
middle_two = numbers[2:4]

print("First three:", first_three)
print("Last three:", last_three)
print("Middle two:", middle_two)
```

### Exercise 5: Modify an Element
Given `fruits = ["apple", "banana", "orange"]`, replace `"banana"` with `"pear"` and print the updated list.

```python
# Exercise 5: Modify an element
fruits = ["apple", "banana", "orange"]
fruits[1] = "pear"

print(fruits)
```

### Exercise 6: Add Elements
Given `fruits = ["apple", "pear"]`:
1. Add `"orange"` to the end.
2. Insert `"banana"` as the second element.
Print the updated list.

```python
# Exercise 6: Add elements
fruits = ["apple", "pear"]

fruits.append("orange")
fruits.insert(1, "banana")

print(fruits)
```
