## Exercise 1

Write a function called `cube` that takes a number and returns its cube (number to the power of 3). Test it with a few values.

```python
def cube(number):
    return number ** 3


# Test the function with a few values
print(cube(2))
print(cube(3))
print(cube(5))
```

## Exercise 2

Use the `ask_ok` function and call it in three different ways:

1. Giving only the mandatory argument.
2. Giving one optional argument.
3. Giving all arguments.

```python
def ask_ok(prompt, retries=4, reminder='Please try again!'):
    while True:
        reply = input(prompt)
        if reply in {'y', 'ye', 'yes'}:
            return True
        if reply in {'n', 'no', 'nop', 'nope'}:
            return False
        retries = retries - 1
        if retries < 0:
            raise ValueError('invalid user response')
        print(reminder)
```

## Exercise 3

Write a function `make_coffee(size="medium", milk=True)` that returns a sentence describing the order, such as `"One medium coffee with milk"`. Test it with no arguments and then by overriding both arguments.

```python
def make_coffee(size="medium", milk=True):
    if milk:
        return f"One {size} coffee with milk"
    return f"One {size} coffee without milk"


# Call with no arguments
print(make_coffee())

# Override both arguments
print(make_coffee("large", False))
```

## Exercise 4

Write a function `average(*numbers)` that returns the average of any amount of numbers passed in. Handle zero arguments gracefully by returning `0`.

```python
def average(*numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)


# Tests
print(average())
print(average(10))
print(average(10, 20, 30))
print(average(5, 10, 15, 20))
```

## Exercise 5

Write a function `full_name(first, last, *, middle=None)` that returns the full name as a string. If `middle` is given, include it in the middle.

```python
def full_name(first, last, *, middle=None):
    if middle is not None:
        return f"{first} {middle} {last}"
    return f"{first} {last}"


# Tests
print(full_name("John", "Smith"))
print(full_name("John", "Smith", middle="Michael"))
```
