## Exercise 9 — `*args`

Write a function `average(*numbers)` that returns the average of any number of numeric arguments.

```python

average(1, 2, 3, 4, 5)
```

What should your function do if no numbers are provided? Handle that case sensibly.


````python
# Your solution here


````

### Solution 9

````python
def average(*numbers):
    if not numbers:
        return None
    return sum(numbers) / len(numbers)

print(average(10, 20))
print(average(1, 2, 3, 4, 5))
print(average())
````

## Exercise 17 — The mutable default argument trap

Run the following:

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("orange"))
```

Explain why the list keeps its previous contents.

Then rewrite the function using `None` as the default value.


````python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("orange"))


# Rewrite the function safely below


````

### Solution 17

````python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("apple"))
print(add_item("banana"))
print(add_item("orange"))

# The default list is created once, when the function is defined.
# It is then reused on subsequent calls.

def add_item_safe(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item_safe("apple"))
print(add_item_safe("banana"))
print(add_item_safe("orange"))
````
