### 5.2.2 Set manipulation

### Exercise 5
Using `python_students` and `java_students`, compute union, intersection, Python-only, and symmetric difference using both operators and method forms.

```python
# Exercise 5
python_students = {"Ana", "Leo", "Kim", "Sam"}
java_students = {"Kim", "Sam", "Tom", "Eva"}

# Union
print("Union (|):", python_students | java_students)
print("Union (.union()):", python_students.union(java_students))

# Intersection
print("Intersection (&):", python_students & java_students)
print("Intersection (.intersection()):", python_students.intersection(java_students))

# Python only
print("Python only (-):", python_students - java_students)
print("Python only (.difference()):", python_students.difference(java_students))

# Exactly one class
print("Symmetric difference (^):", python_students ^ java_students)
print("Symmetric difference (.symmetric_difference()):",
      python_students.symmetric_difference(java_students))
```

### Exercise 6
Check whether `"Tom"` is in `python_students` and whether `python_students` is a subset of the union.

```python
# Exercise 6
all_students = python_students | java_students

print("Tom in Python:", "Tom" in python_students)
print("Subset using .issubset():", python_students.issubset(all_students))
print("Subset using <=:", python_students <= all_students)
```

### Exercise 7
Add `"Nina"` to `python_students`. Demonstrate `.remove()` and `.discard()`, including their different behavior when an element is missing.

```python
# Exercise 7
python_students.add("Nina")
python_students.remove("Leo")

print("After adding Nina and removing Leo:", python_students)

# remove() raises KeyError if the item is missing.
try:
    python_students.remove("Leo")
except KeyError:
    print("remove('Leo'): KeyError because Leo is no longer in the set.")

# discard() does not raise an error if the item is missing.
python_students.discard("Leo")
print("discard('Leo'): no error even though Leo is missing.")
```

Print the price of `"banana"` using `[]` indexing and `.get()`.

```python
# Exercise 3
prices = {"apple": 0.5, "banana": 0.3, "cherry": 2.0}

print(prices["banana"])
print(prices.get("banana"))
```

Try `prices["mango"]`, `prices.get("mango")`, and `prices.get("mango", "not sold here")`. The first expression is commented out so the notebook can continue running.

### 3.3.2 Iterating over dictionaries

### Exercise 6
Print `f"{fruit}: ${price}"` for every item using `.items()`.

```python
# Exercise 6
for fruit, price in prices.items():
    print(f"{fruit}: ${price}")
```

### Exercise 7
Print just the keys using `.keys()`, then again without a method.

```python
# Exercise 7
print("Using .keys():")
for fruit in prices.keys():
    print(fruit)

print("\nWithout a method:")
for fruit in prices:
    print(fruit)

print("\nA dictionary iterates over its keys by default.")
```

### Exercise 8
Print just the values and compute the total using `sum(prices.values())`.

```python
# Exercise 8
for price in prices.values():
    print(price)

total = sum(prices.values())
print("Total:", total)
```

### Exercise 9
Convert `prices.items()` into a list of tuples, sort it by price, and print the cheapest fruit.

```python
# Exercise 9
items = list(prices.items())
sorted_items = sorted(items, key=lambda item: item[1])

print("Items as a list of tuples:", items)
print("Sorted by price:", sorted_items)
print("Cheapest fruit:", sorted_items[0][0])
```
